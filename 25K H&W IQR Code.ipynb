```python
import pandas as pds
import matplotlib.pyplot as plt
import numpy as np
file = ('SOCR-HeightWeight.csv')
newData=pds.read_csv(file)
newData.head()

Q1H = newData["Height(Inches)"].quantile(0.25)
Q2H = newData["Height(Inches)"].quantile(0.5)
Q3H = newData["Height(Inches)"].quantile(0.75)

IQRH = (Q3H - Q1H).round(2)

print("Q1H:",Q1H)
print("Q2H:",Q2H)
print("Q3H:",Q3H)
print("Interquartile Range for Height:",IQRH)
print(" ")


Q1W = newData["Weight(Pounds)"].quantile(0.25)
Q2W = newData["Weight(Pounds)"].quantile(0.5)
Q3W = newData["Weight(Pounds)"].quantile(0.75)

IQRW = (Q3W - Q1W).round(2)
print("Q1W:",Q1W)
print("Q2W:",Q2W)
print("Q3W:",Q3W)
print("Interquartile Range for Weight:",IQRW)
print(" ")

MaxH=newData.max()
print(MaxH)

fig = plt.figure(figsize =(10, 10))
plt.boxplot(newData["Height(Inches)"], orientation = "horizontal")
plt.show()



fig = plt.figure(figsize =(10, 10))
plt.boxplot(newData["Weight(Pounds)"], orientation = "horizontal")
plt.show()
```
